```python
# Import necessary libraries and modules

from sklearn.datasets import load_iris
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load the Iris dataset from scikit-learn
iris = load_iris()

# Convert the Iris dataset (Bunch object) into a pandas DataFrame.
df = pd.DataFrame(
    iris.data,
    columns= iris.feature_names
)

# Add a 'species' column to the DataFrame, derived from the target variable.
df["species"] = iris.target
```

```python
# Display the first 5 rows of the DataFrame.
df.head()
```

```python
# Display concise summary of the DataFrame, including data types and non-null values.
df.info()
```

```python
# Generate descriptive statistics of the DataFrame.
df.describe()
```

```python
# Group by species and calculate the mean of numerical columns.
from pandas.core import groupby
df.groupby('species').mean()
```

```python
# This cell visualizes the relationship between petal length and petal width, colored by species.
plt.scatter(
    df['petal length (cm)'],
    df['petal width (cm)'],
    c= df['species']
)

plt.xlabel('Petal length (cm)')
plt.ylabel('Petal width (cm)')
plt.title("Petal length vs Petal width ")

plt.show()
```
